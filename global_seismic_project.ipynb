```python
import requests
import pandas as pd
from datetime import datetime

url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

all_records = []
start_year = datetime.now().year - 5   # last 5 years
end_year = datetime.now().year

for year in range(start_year, end_year + 1):
    for month in range(1, 13):
        start_date = f"{year}-{month:02d}-01"
        if month == 12:
            end_date = f"{year+1}-01-01"
        else:
            end_date = f"{year}-{month+1:02d}-01"

        params = {
            "format": "geojson",
            "starttime": start_date,
            "endtime": end_date,
            "minmagnitude": 3
        }

        response = requests.get(url, params=params)
        if response.status_code != 200:
            print(f"⚠️ Failed for {start_date}: {response.text[:200]}")
            continue

        try:
            data = response.json()
        except Exception as e:
            print(f"⚠️ JSON error for {start_date}: {e}")
            continue

        for f in data["features"]:
            p = f["properties"]
            g = f["geometry"]["coordinates"]
            all_records.append({
                "id": f.get("id"),
                "time": pd.to_datetime(p.get("time"), unit="ms"),
                "updated": pd.to_datetime(p.get("updated"), unit="ms"),
                "latitude": g[1] if g else None,
                "longitude": g[0] if g else None,
                "depth_km": g[2] if g else None,
                "mag": p.get("mag"),
                "magType": p.get("magType"),
                "place": p.get("place"),
                "status": p.get("status"),
                "tsunami": p.get("tsunami"),
                "alert": p.get("alert"),
                "felt": p.get("felt"),
                "cdi": p.get("cdi"),
                "mmi": p.get("mmi"),
                "sig": p.get("sig"),
                "net": p.get("net"),
                "code": p.get("code"),
                "ids": p.get("ids"),
                "sources": p.get("sources"),
                "types": p.get("types"),
                "nst": p.get("nst"),
                "dmin": p.get("dmin"),
                "rms": p.get("rms"),
                "gap": p.get("gap"),
                "type": p.get("type"),
                "magError": p.get("magError"),
                "depthError": p.get("depthError" ),
                "magNst": p.get("magNst"),
                "magSource": p.get("magSource")
            })







df = pd.DataFrame(all_records)
```

**NULL VALUES HANDLING**

```python
df[['felt','cdi','mmi']]=df[['felt','cdi','mmi']].fillna(0)
```

```python
cols=['nst','dmin','rms','gap']
```

```python
for col in cols:
  df[col]=df[col].fillna(df[col].median())
```

```python
def get_alert(mag):
  if mag <5.5:
    return 'green'
  elif 5.5<=mag<6.5:
    return 'yellow'
  elif 6.5<=mag<7.5:
    return 'orange'
  else:
    return 'red'
```

```python
df['alert']=df['alert'].fillna(df['mag'].apply(get_alert))
```

```python
df['year']=df['time'].dt.year
df['month']=df['time'].dt.month_name()
df['day']=df['time'].dt.day
df['day_of_week']=df['time'].dt.day_name()
df['hour']=df['time'].dt.hour
```

```python
df
```

```python
def get_depth_catagory(depth):
  if depth <=70:
    return 'shallow'
  elif depth <=300:
    return 'intermediate'
  else:
    return 'deep'
```

```python

df['depth_catagory']=df['depth_km'].apply(get_depth_catagory)
```

```python
def mag_flag(mag):
  if mag>=7.0:
    return 'Destructive'
  elif mag>=6.0:
    return 'Storng'
  else:
    return 'Normal'
```

```python
df['mag_type_flag']=df['mag'].apply(mag_flag)
```

```python
cat_col=df.select_dtypes(include="object").columns
```

```python
for col in cat_col:
  df[col]=df[col].astype(str).str.lower().str.strip()
```

```python
import reverse_geocoder as rg
coords = list(zip(df['latitude'], df['longitude']))
results = rg.search(coords)
df['country_code'] = [res['cc'] for res in results]
```

```python
df
```

```python
import pycountry
```

```python
def get_country_name(code):
    try:
        return pycountry.countries.get(alpha_2=code.upper()).name.lower()
    except:
        return 'unknown'

```

```python
df['country'] = df['country_code'].apply(get_country_name)
```
