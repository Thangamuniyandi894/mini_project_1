```python
# 7. Month with highest number of earthquakes.

qus7="""select month,count(*) AS total_counts
from earthquakes 
group by month
order by total_counts DESC 
limit 1;"""
df7=pd.read_sql(qus7,conn)
df7
```

```python
# 8. Day of week with most earthquakes.

qus8="""select day_of_week,count(*) as total_earthquakes 
from earthquakes 
group by day_of_week 
order by total_earthquakes desc
limit 1;"""
df8=pd.read_sql(qus8,conn)
df8
```

```python
#21.Find the top 5 countries with the highest average magnitude of earthquakes in the past 5 years  
qus21="""select country, avg(mag) as avg_mag 
from earthquakes 
where year>=(select max(year)-5 from earthquakes)
group by country
order by avg_mag desc 
limit 5;"""
df21=pd.read_sql(qus21,conn)
df21
```

```python
# 22.Find countries that have experienced both shallow and deep earthquakes within the same month.
qus22="""select country, year,month 
from earthquakes 
group by country, year, month 
having sum(case when depth_km< 70 then 1 else 0 end) > 0 
   and sum(case when depth_km > 300 then 1 else 0 end) > 0;"""
df22=pd.read_sql(qus22,conn)
df22
```

```python
# 23.Compute the year-over-year growth rate in the total number of earthquakes globally.

qus23="""with yearlycounts as (
    select year as eq_year, count(*) as total_eq 
    from earthquakes 
    group by year
)
select eq_year, total_eq,
    lag(total_eq) over (order by eq_year) as prev_year_total,
    round(((total_eq - lag(total_eq) over (order by eq_year)) / lag(total_eq) over (order by eq_year)) * 100, 2) as yoy_growth_percent
from yearlyCounts;"""
df23=pd.read_sql(qus23,conn)
df23
```

```python
#24. List the 3 most seismically active regions by combining both frequency and average magnitude.
qus24="""select place, count(*) as frequency, avg(mag) as avg_mag 
from earthquakes 
group by place 
order by (count(*) * avg(mag)) desc 
limit 3;"""
df24=pd.read_sql(qus24,conn)
df24
```

```python
#  25. For each country, calculate the average depth of earthquakes within ±5° latitude range of the equator.
qus25="""select country, avg(depth_km) as avg_depth 
from earthquakes 
where latitude between -5 and 5 
group by country;"""
df25=pd.read_sql(qus25,conn)
df25
```

```python
# 26. Identify countries having the highest ratio of shallow to deep earthquakes.
qus26="""select country, 
    sum(case when depth_km < 70 then 1 else 0 end) / 
    nullif(sum(case when depth_km > 300 then 1 else 0 end), 0) as shallow_to_deep_ratio 
from earthquakes 
group by country 
order by shallow_to_deep_ratio desc;"""
df26=pd.read_sql(qus26,conn)
df26
```

```python
#27. Find the average magnitude difference between earthquakes with tsunami alerts and those without.
qus27="""select (select avg(mag) from earthquakes where tsunami=1)-(
select avg(mag) from earthquakes where tsunami=0) as avg_magnitude_diff;"""
df27=pd.read_sql(qus27,conn)
df27
```

```python
#28. Using the gap and rms columns, identify events with the lowest data reliability (highest average error margins).
qus28="""select *, (gap + rms) as total_error_margin 
from earthquakes 
order by total_error_margin desc 
limit 10;"""
df28=pd.read_sql(qus28,conn)
df28
```
